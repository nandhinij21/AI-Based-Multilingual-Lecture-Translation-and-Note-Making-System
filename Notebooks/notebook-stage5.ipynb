```python
import os
import json
import asyncio
import subprocess
import re
import librosa
import numpy as np
from pathlib import Path
import nest_asyncio

# Apply the asyncio patch for interactive notebook execution kernels
nest_asyncio.apply()

# ── CONFIGURATION & PATHS ───────────────────────────────────────────────────
TRANSLATION_FILE = "/kaggle/input/datasets/darshanchoudhary029/translated/Tu1PBgkHzvk_large-v3-turbo_translated.json"
VIDEO_NO_AUDIO = "video/no_audio.mp4"
ORIGINAL_VIDEO = "video/yRpLlJmRo2w.mp4"
SEGMENT_DIR = Path("/kaggle/working/segments")
DUBBED_DIR = Path("/kaggle/working/dubbed")

VOICE_PROFILE_MAP = {
    "tam_taml": {"male": "ta-IN-ValluvarNeural", "female": "ta-IN-PallaviNeural"},
    "tel_telu": {"male": "te-IN-MohanNeural",    "female": "te-IN-ShrutiNeural"},
    "mar_deva": {"male": "mr-IN-ManoharNeural",  "female": "mr-IN-AarohiNeural"},
    "ben_beng": {"male": "bn-IN-BashkarNeural",  "female": "bn-IN-TanishaaNeural"}
}

MAX_CONCURRENT_REQUESTS = 10

# ── AUTOMATED GENDER DETECTION DETECTOR ──────────────────────────────────────

GENDER_OVERRIDE = "auto" 

def detect_speaker_gender(video_path):
    """Samples multiple timestamps in the video midsection to calculate a true median voice pitch."""
    if GENDER_OVERRIDE.lower() in ["male", "female"]:
        print(f"⚙️ Manual Override Active: Forcing [{GENDER_OVERRIDE.upper()}] voice profiles.")
        return GENDER_OVERRIDE.lower()

    temp_sample_wav = "/kaggle/working/gender_sample_robust.wav"
    pitches_collected = []
    
    # Scan 3 distinct time slices deep in the lecture (at 2, 5, and 8 minutes) to dodge intro music
    sample_timestamps = ["120", "300", "480"]
    
    print("🔍 Extracting multi-point audio samples for stable voice classification...")
    for timestamp in sample_timestamps:
        try:
            subprocess.run([
                "ffmpeg", "-y", "-ss", timestamp, "-i", video_path, 
                "-t", "10", "-ar", "16000", "-ac", "1", temp_sample_wav
            ], check=True, capture_output=True)
            
            if os.path.exists(temp_sample_wav):
                y, sr = librosa.load(temp_sample_wav, sr=16000)
                f0, voiced_flag, voiced_probs = librosa.pyin(
                    y, fmin=librosa.note_to_hz('C2'), fmax=librosa.note_to_hz('C7')
                )
                clean_pitches = f0[~np.isnan(f0)]
                pitches_collected.extend(clean_pitches)
        except Exception:
            continue

    if os.path.exists(temp_sample_wav):
        os.remove(temp_sample_wav)

    if len(pitches_collected) == 0:
        print("⚠️ Voice metrics flatlined or silent. Defaulting to MALE profile.")
        return "male"
        
    final_median_pitch = np.median(pitches_collected)
    print(f"📊 Robust Averaged Speaker Pitch: {final_median_pitch:.1f} Hz")
    
    # Clear mathematical line between male/female vocal chords
    if final_median_pitch > 165.0:
        print("👩 Classifier Decision: FEMALE Speaker Detected.")
        return "female"
    else:
        print("👨 Classifier Decision: MALE Speaker Detected.")
        return "male"

# ── TEXT HANDLING HELPERS ───────────────────────────────────────────────────

def clean_tts_text(text):
    text = re.sub(r'<[^>]*>', '', text)
    for char in ['{', '}', '[', ']', '"', "'", '\\', '*', '_', '-']:
        text = text.replace(char, '')
    return text.strip()

def split_into_sentences(text):
    sentences = re.split(r'(?<=[.!?।])\s+', text)
    return [s.strip() for s in sentences if len(s.strip()) > 1]

def create_silence(duration, out_path):
    if duration <= 0:
        duration = 0.1
    subprocess.run([
        "ffmpeg", "-y", "-f", "lavfi", 
        "-i", "anullsrc=r=24000:cl=stereo", 
        "-t", f"{duration:.3f}", out_path
    ], check=True, capture_output=True)

def get_audio_duration(file_path):
    result = subprocess.run([
        "ffprobe", "-v", "error", "-show_entries", "format=duration",
        "-of", "default=noprint_wrappers=1:nokey=1", file_path
    ], capture_output=True, text=True, check=True)
    return float(result.stdout.strip())

# ── ASYNCHRONOUS NATURAL VOICE DOWNLOAD WORKER ──────────────────────────────

async def download_segment_task(i, text, voice, seg_wav, target_duration, semaphore):
    """Downloads voice files at 1.0x natural speed to avoid robotic glitches."""
    import edge_tts
    async with semaphore:
        if not text or target_duration <= 0:
            create_silence(max(target_duration, 0.1), seg_wav)
            return
            
        raw_mp3 = seg_wav.replace(".wav", "_raw.mp3")
        try:
            communicate = edge_tts.Communicate(text, voice)
            await communicate.save(raw_mp3)
            
            if os.path.exists(raw_mp3) and os.path.getsize(raw_mp3) > 100:
                # Convert raw mono to standard high-quality stereo channel mapping smoothly
                subprocess.run([
                    "ffmpeg", "-y", "-i", raw_mp3, 
                    "-ar", "24000", "-ac", "2", seg_wav
                ], check=True, capture_output=True)
                os.remove(raw_mp3)
            else:
                create_silence(target_duration, seg_wav)
        except Exception:
            create_silence(target_duration, seg_wav)

# ── PIPELINE STREAM ENGINE ───────────────────────────────────────────────────

async def run_pipeline_async():
    DUBBED_DIR.mkdir(parents=True, exist_ok=True)
    SEGMENT_DIR.mkdir(parents=True, exist_ok=True)
    
    try:
        video_duration = get_audio_duration(VIDEO_NO_AUDIO)
        print(f"🎬 Universal Video Track Length Detected: {video_duration:.2f} seconds")
    except Exception as e:
        print(f"⚠️ Error reading video length ({e}). Falling back to safe default sizing.")
        video_duration = 1126.0
        
    detected_gender = detect_speaker_gender(ORIGINAL_VIDEO)
    print(f"👩‍🏫 Selected Global Gender Profile: {detected_gender.upper()}")
    
    print(f"📖 Reading translation file structure...")
    with open(TRANSLATION_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)
        
    translations_block = data.get("translations", {})
    semaphore = asyncio.Semaphore(MAX_CONCURRENT_REQUESTS)
    
    for json_key, content in translations_block.items():
        normalized_key = json_key.lower()
        
        if normalized_key in VOICE_PROFILE_MAP:
            voice_profile = VOICE_PROFILE_MAP[normalized_key][detected_gender]
            paragraph_text = content.get("translated_text", "")
            segments_list = content.get("segments", [])
            
            if not paragraph_text or not segments_list:
                continue
                
            print(f"\n🚀 Processing language channel: {json_key} ── Using Voice: [{voice_profile}]")
            
            seg_subdir = SEGMENT_DIR / json_key
            seg_subdir.mkdir(parents=True, exist_ok=True)
            
            out_wav = f"{DUBBED_DIR}/{json_key}_dubbed.wav"
            out_mp4 = f"{DUBBED_DIR}/{json_key}_dubbed.mp4"
            
            sentences = split_into_sentences(paragraph_text)
            total_sentences = len(sentences)
            
            file_list = []
            download_tasks = []
            prev_end = 0.0
            
            for i, seg in enumerate(segments_list):
                start = float(seg.get("start", 0.0))
                end = float(seg.get("end", start + 1.0))
                duration = end - start
                
                text_payload = clean_tts_text(sentences[i]) if i < total_sentences else ""
                
                gap = start - prev_end
                if gap > 0.01:
                    silence_path = str(seg_subdir / f"silence_{i:04d}.wav")
                    create_silence(gap, silence_path)
                    file_list.append(silence_path)
                
                seg_wav = str(seg_subdir / f"voice_{i:04d}.wav")
                file_list.append(seg_wav)
                
                download_tasks.append(
                    download_segment_task(i, text_payload, voice_profile, seg_wav, duration, semaphore)
                )
                prev_end = end
            
            print(f"⚡ Downloading voice segments at 1.0x natural tempo...")
            await asyncio.gather(*download_tasks)
            
            concat_manifest = f"/kaggle/working/concat_{json_key}.txt"
            with open(concat_manifest, "w") as f_manifest:
                for path in file_list:
                    if os.path.exists(path):
                        f_manifest.write(f"file '{path}'\n")
            
            # Stitch raw segments together
            temp_stitched_wav = f"/kaggle/working/temp_{json_key}_stitched.wav"
            print(f"🎛️ Creating uniform master audio stream...")
            subprocess.run([
                "ffmpeg", "-y", "-f", "concat", "-safe", "0", 
                "-i", concat_manifest, "-b:a", "192k", temp_stitched_wav
            ], check=True, capture_output=True)
            
            if os.path.exists(concat_manifest):
                os.remove(concat_manifest)
                
            if os.path.exists(temp_stitched_wav):
                # Calculate global tempo scale factor to fit the target video duration perfectly
                stitched_duration = get_audio_duration(temp_stitched_wav)
                global_speed = stitched_duration / video_duration
                
                # Clip extreme bounds to keep voice sounding natural and clear
                global_speed = max(0.85, min(global_speed, 1.35))
                print(f"📈 Harmonizing absolute timeline synchronization factor: {global_speed:.3f}x")
                
                # Apply a smooth, uniform time-stretch filter to the entire continuous file
                subprocess.run([
                    "ffmpeg", "-y", "-i", temp_stitched_wav, 
                    "-filter:a", f"atempo={global_speed:.3f}", 
                    "-t", f"{video_duration:.3f}", "-ac", "2", out_wav
                ], check=True, capture_output=True)
                
                os.remove(temp_stitched_wav)
                
                # Merge perfectly synchronized stereo audio into the final MP4 container
                subprocess.run([
                    "ffmpeg", "-y", "-i", VIDEO_NO_AUDIO, "-i", out_wav, 
                    "-map", "0:v:0", "-map", "1:a:0", "-c:v", "copy", "-c:a", "aac", "-ac", "2", out_mp4
                ], check=True, capture_output=True)
                
                print(f"🎉 Production Completed Successfully! Flawless real-time matching for: {json_key}")

await run_pipeline_async()
```
